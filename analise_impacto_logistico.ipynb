```python
import torch
import torch.nn as nn

# 1. Definindo a arquitetura com Forward Pass explícito
class NeuronioForward(nn.Module):
    def __init__(self, quantidade_entradas):
        super(NeuronioForward, self).__init__()
        # Camada linear: calcula (X * W) + b
        self.camada_linear = nn.Linear(quantidade_entradas, 1)
        # Função de ativação
        self.ativacao = nn.Sigmoid()

    def forward(self, x):
        # Passo 1 do Forward Pass: Soma Ponderada + Viés
        soma = self.camada_linear(x)
        # Passo 2 do Forward Pass: Aplicação da Ativação (Sigmoide)
        saida = self.ativacao(soma)
        return soma, saida

# 2. Instanciando o modelo
modelo = NeuronioForward(quantidade_entradas=3)

# 3. Criando o Tensor de Entrada (3 atributos para 2 amostras)
entradas = torch.tensor([
    [0.85, 0.30, 0.95],
    [0.10, 0.15, 0.20]
], dtype=torch.float32)

# 4. Executando o Forward Pass
soma_bruta, resultado_final = modelo(entradas)

# 5. Exibindo os passos intermediários do Forward Pass
print("--- EXECUÇÃO DO FORWARD PASS ---")
print("1. Entradas (X):\n", entradas)
print("\n2. Soma Ponderada Brutal [Z = (X * W) + b]:\n", soma_bruta.detach())
print("\n3. Saída Final Ativada [A = Sigmoid(Z)]:\n", resultado_final.detach())
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Definição do Modelo de Classificação de Crédito
class ClassificadorBomPagador(nn.Module):
    def __init__(self, quantidade_atributos):
        super(ClassificadorBomPagador, self).__init__()
        # Camada linear com 3 entradas e 1 saída
        self.camada_linear = nn.Linear(quantidade_atributos, 1)
        # Função de ativação Sigmoide (retorna probabilidade entre 0 e 1)
        self.ativacao = nn.Sigmoid()

    def forward(self, x):
        z = self.camada_linear(x)
        probabilidade = self.ativacao(z)
        return probabilidade

# 2. Dados de Exemplo (Normalizados entre 0 e 1)
# Atributos: [Renda Mensal, Relação Dívida/Renda, Histórico de Pontualidade]
X_clientes = torch.tensor([
    [0.85, 0.15, 0.95],  # Cliente 1: Renda alta, baixa dívida, ótimo histórico -> (Bom Pagador)
    [0.20, 0.80, 0.30],  # Cliente 2: Renda baixa, alta dívida, histórico ruim -> (Mau Pagador)
    [0.70, 0.35, 0.80],  # Cliente 3: Perfil equilibrado -> (Bom Pagador)
    [0.30, 0.70, 0.20]   # Cliente 4: Alto risco -> (Mau Pagador)
], dtype=torch.float32)

# Rótulos Reais (1 = Bom Pagador, 0 = Mau Pagador)
y_real = torch.tensor([[1.0], [0.0], [1.0], [0.0]], dtype=torch.float32)

# 3. Configuração do Modelo, Perda e Otimizador
modelo = ClassificadorBomPagador(quantidade_atributos=3)
criterio = nn.BCELoss()  # Binary Cross Entropy Loss
otimizador = optim.SGD(modelo.parameters(), lr=0.2)

# 4. Treinamento da Rede Neural
print("--- TREINANDO O CLASSIFICADOR DE CRÉDITO ---")
for epoca in range(1, 201):
    previsoes = modelo(X_clientes)
    perda = criterio(previsoes, y_real)

    otimizador.zero_grad()
    perda.backward()
    otimizador.step()

    if epoca % 50 == 0:
        print(f"Época {epoca}/200 - Perda (Loss): {perda.item():.4f}")

# 5. Avaliação dos Resultados nos Clientes
print("\n--- AVALIAÇÃO DE ANÁLISE DE CRÉDITO ---")
with torch.no_grad():
    probabilidades = modelo(X_clientes)
    for i, prob in enumerate(probabilidades):
        prob_val = prob.item()
        status = "BOM PAGADOR ✅" if prob_val >= 0.5 else "ALTO RISCO / MAU PAGADOR ❌"
        print(f"Cliente {i+1}: Probabilidade = {prob_val:.2%} | Decisão = {status}")
```

```python
import math


def sigmoide(x):
    return 1 / (1 + math.exp(-x))


# 1. Entradas (ex: duas características de um dado)
entradas = [2.0, 3.0]

# 2. Pesos e Viés (iniciados com valores genéricos)
pesos = [0.8, -0.5]
vies = 0.1

# 3. Soma Ponderada: (2.0 * 0.8) + (3.0 * -0.5) + 0.1
soma_ponderada = sum(e * p for e, p in zip(entradas, pesos)) + vies

# 4. Saída do Neurônio (passada pela função de ativação)
saida = sigmoide(soma_ponderada)

print(f"Soma ponderada: {soma_ponderada}")
print(f"Saida do neuronio: {saida:.4f}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Definição da estrutura do Neurônio
class ClassificadorCredito(nn.Module):
    def __init__(self, entradas):
        super(ClassificadorCredito, self).__init__()
        self.camada_linear = nn.Linear(entradas, 1)
        self.ativacao = nn.Sigmoid()

    def forward(self, x):
        z = self.camada_linear(x)
        return self.ativacao(z)

# 2. Dados dos Clientes: [Renda, Dívida, Histórico]
X_clientes = torch.tensor([
    [0.85, 0.15, 0.95],  # Cliente 1: Renda alta, dívida baixa, ótimo histórico -> Bom
    [0.20, 0.80, 0.30],  # Cliente 2: Renda baixa, dívida alta, histórico ruim -> Mau
    [0.70, 0.35, 0.80],  # Cliente 3: Perfil equilibrado -> Bom
    [0.30, 0.70, 0.20]   # Cliente 4: Alto risco -> Mau
], dtype=torch.float32)

# Respostas esperadas (1 = Bom Pagador, 0 = Mau Pagador)
y_real = torch.tensor([[1.0], [0.0], [1.0], [0.0]], dtype=torch.float32)

# 3. Configuração do Modelo, Função de Perda e Otimizador
modelo = ClassificadorCredito(entradas=3)
criterio = nn.BCELoss()  # Erro por Entropia Cruzada
otimizador = optim.SGD(modelo.parameters(), lr=0.2)  # Gradiente Descendente

# 4. Loop de Aprendizado (Treinamento por 200 épocas)
print("--- INICIANDO TREINAMENTO ---")
for epoca in range(1, 201):
    previsoes = modelo(X_clientes)        # 1. Previsão (Forward)
    perda = criterio(previsoes, y_real)   # 2. Cálculo do Erro (Loss)

    otimizador.zero_grad()               # Limpa gradientes anteriores
    perda.backward()                     # 3. Retropropagação (Backpropagation)
    otimizador.step()                    # 4. Ajuste dos Pesos (Step)

    if epoca % 50 == 0:
        print(f"Época {epoca}/200 - Perda (Loss): {perda.item():.4f}")

# 5. Avaliação Final nos Clientes
print("\n--- RESULTADO FINAL ---")
with torch.no_grad():
    probabilidades = modelo(X_clientes)
    for i, prob in enumerate(probabilidades):
        status = "BOM PAGADOR ✅" if prob.item() >= 0.5 else "MAU PAGADOR / RISCO ❌"
        print(f"Cliente {i+1}: Probabilidade = {prob.item():.2%} | Status = {status}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Arquitetura da Rede Neural Multicamadas (MLP)
class RedeNeuralCredito(nn.Module):
    def __init__(self, entradas):
        super(RedeNeuralCredito, self).__init__()
        # Camada Oculta: recebe as 3 entradas e passa para 4 neurônios internos
        self.camada_oculta = nn.Linear(entradas, 4)
        self.relu = nn.ReLU() # Função de Ativação não-linear para a camada oculta

        # Camada de Saída: recebe os 4 neurônios internos e reduz para 1 probabilidade final
        self.camada_saida = nn.Linear(4, 1)
        self.sigmoide = nn.Sigmoid()

    def forward(self, x):
        # Passo 1: Passa pela camada oculta + ReLU
        x = self.relu(self.camada_oculta(x))
        # Passo 2: Passa pela camada de saída + Sigmoide
        x = self.sigmoide(self.camada_saida(x))
        return x

# 2. Reutilizando os dados de clientes
X_clientes = torch.tensor([
    [0.85, 0.15, 0.95],  # Cliente 1: Bom
    [0.20, 0.80, 0.30],  # Cliente 2: Mau
    [0.70, 0.35, 0.80],  # Cliente 3: Bom
    [0.30, 0.70, 0.20]   # Cliente 4: Mau
], dtype=torch.float32)

y_real = torch.tensor([[1.0], [0.0], [1.0], [0.0]], dtype=torch.float32)

# 3. Instanciando o novo modelo, erro e otimizador
rede_modelo = RedeNeuralCredito(entradas=3)
criterio = nn.BCELoss()
otimizador = optim.Adam(rede_modelo.parameters(), lr=0.01) # Usando o otimizador Adam

# 4. Loop de Treinamento
print("--- TREINANDO A REDE NEURAL MULTICAMADAS ---")
for epoca in range(1, 301):
    previsoes = rede_modelo(X_clientes)
    perda = criterio(previsoes, y_real)

    otimizador.zero_grad()
    perda.backward()
    otimizador.step()

    if epoca % 100 == 0:
        print(f"Época {epoca}/300 - Perda (Loss): {perda.item():.4f}")

# 5. Testando a Rede em Novos Clientes
novos_clientes = torch.tensor([
    [0.90, 0.10, 0.90],  # Cliente A (Perfil ótimo)
    [0.15, 0.90, 0.10],  # Cliente B (Risco alto)
    [0.60, 0.40, 0.60]   # Cliente C (Perfil mediano)
], dtype=torch.float32)

print("\n--- TESTANDO A REDE EM NOVOS CLIENTES ---")
with torch.no_grad():
    probabilidades = rede_modelo(novos_clientes)
    for i, prob in enumerate(probabilidades):
        letra = chr(65 + i)
        status = "BOM PAGADOR ✅" if prob.item() >= 0.5 else "MAU PAGADOR / RISCO ❌"
        print(f"Novo Cliente {letra}: Probabilidade = {prob.item():.2%} | Status = {status}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Vocabulário: Mapeando cada palavra para um número (Tokenização)
vocabulario = {
    "<PAD>": 0, "excelente": 1, "horrivel": 2, "bom": 3,
    "ruim": 4, "produto": 5, "servico": 6, "muito": 7
}

# 2. Dataset simples: Frases representadas por números e seus rótulos (1 = Positivo, 0 = Negativo)
# Frase 1: "produto excelente" -> [5, 1] (Positivo)
# Frase 2: "servico horrivel"  -> [6, 2] (Negativo)
# Frase 3: "muito bom"          -> [7, 3] (Positivo)
# Frase 4: "muito ruim"         -> [7, 4] (Negativo)

X_texto = torch.tensor([
    [5, 1],
    [6, 2],
    [7, 3],
    [7, 4]
], dtype=torch.long)

y_sentimento = torch.tensor([[1.0], [0.0], [1.0], [0.0]], dtype=torch.float32)

# 3. Criando a Rede Neural com camada de Embedding
class ClassificadorTexto(nn.Module):
    def __init__(self, tamanho_vocab, dim_embedding):
        super(ClassificadorTexto, self).__init__()
        # Embedding: Transforma números em vetores densos de significados
        self.embedding = nn.Embedding(tamanho_vocab, dim_embedding)
        self.camada_oculta = nn.Linear(dim_embedding, 4)
        self.relu = nn.ReLU()
        self.saida = nn.Linear(4, 1)
        self.sigmoide = nn.Sigmoid()

    def forward(self, x):
        # 1. Transforma palavras em vetores e tira a média da frase
        embeds = self.embedding(x)
        vetor_frase = embeds.mean(dim=1)

        # 2. Passa pelas camadas neurais
        x = self.relu(self.camada_oculta(vetor_frase))
        x = self.sigmoide(self.saida(x))
        return x

# 4. Instanciando e Treinando
modelo_nlp = ClassificadorTexto(tamanho_vocab=len(vocabulario), dim_embedding=8)
criterio = nn.BCELoss()
otimizador = optim.Adam(modelo_nlp.parameters(), lr=0.05)

print("--- TREINANDO MODELO DE NLP ---")
for epoca in range(1, 201):
    previsoes = modelo_nlp(X_texto)
    perda = criterio(previsoes, y_sentimento)

    otimizador.zero_grad()
    perda.backward()
    otimizador.step()

    if epoca % 50 == 0:
        print(f"Época {epoca}/200 - Perda: {perda.item():.4f}")

# 5. Testando em novas frases:
# Nova Frase A: "bom produto" -> [3, 5]
# Nova Frase B: "ruim servico" -> [4, 6]
novas_frases = torch.tensor([
    [3, 5],
    [4, 6]
], dtype=torch.long)

print("\n--- AVALIANDO SENTIMENTO DE NOVAS FRASES ---")
with torch.no_grad():
    resul = modelo_nlp(novas_frases)
    print(f"Frase 'bom produto': {resul[0].item():.2%} -> {'POSITIVA ✅' if resul[0] >= 0.5 else 'NEGATIVA ❌'}")
    print(f"Frase 'ruim servico': {resul[1].item():.2%} -> {'POSITIVA ✅' if resul[1] >= 0.5 else 'NEGATIVA ❌'}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import re

# 1. Conjunto de Dados Prático em Português
dados_treino = [
    ("Este produto é maravilhoso, adorei muito", 1),
    ("Excelente atendimento e entrega super rápida", 1),
    ("Muito bom, superou minhas expectativas", 1),
    ("Gostei bastante, recomendo a todos", 1),
    ("Péssimo produto, veio quebrado e não funciona", 0),
    ("Serviço horrível, nunca mais compro aqui", 0),
    ("Muito ruim, odiei a experiência", 0),
    ("Atrasou demais e o atendimento foi péssimo", 0)
]

# Função para limpar o texto (remover pontuação e caixa alta)
def limpar_texto(texto):
    texto = texto.lower()
    texto = re.sub(r'[^\w\s]', '', texto)
    return texto.split()

# 2. Construindo o Vocabulário
vocabulario = {"<PAD>": 0, "<UNK>": 1}
for frase, _ in dados_treino:
    palavras = limpar_texto(frase)
    for palavra in palavras:
        if palavra not in vocabulario:
            vocabulario[palavra] = len(vocabulario)

print(f"Tamanho do vocabulário criado: {len(vocabulario)} palavras")

# Converter frase para lista de números (tokens)
def frase_para_tokens(frase, max_len=6):
    palavras = limpar_texto(frase)
    tokens = [vocabulario.get(p, vocabulario["<UNK>"]) for p in palavras]
    # Padding / Truncamento para manter tamanho fixo
    if len(tokens) < max_len:
        tokens += [vocabulario["<PAD>"]] * (max_len - len(tokens))
    else:
        tokens = tokens[:max_len]
    return tokens

# Preparando os tensores de treino
X_treino = torch.tensor([frase_para_tokens(f) for f, _ in dados_treino], dtype=torch.long)
y_treino = torch.tensor([[rotulo] for _, rotulo in dados_treino], dtype=torch.float32)

# 3. Modelo de Classificação de Sentimentos
class ClassificadorSentimento(nn.Module):
    def __init__(self, tam_vocab, dim_embed):
        super(ClassificadorSentimento, self).__init__()
        self.embedding = nn.Embedding(tam_vocab, dim_embed, padding_idx=0)
        self.fc1 = nn.Linear(dim_embed, 8)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(8, 1)
        self.sigmoide = nn.Sigmoid()

    def forward(self, x):
        embeds = self.embedding(x).mean(dim=1)
        x = self.relu(self.fc1(embeds))
        x = self.sigmoide(self.fc2(x))
        return x

modelo_nlp = ClassificadorSentimento(len(vocabulario), dim_embed=16)
criterio = nn.BCELoss()
otimizador = optim.Adam(modelo_nlp.parameters(), lr=0.02)

# 4. Treinamento
print("\n--- TREINANDO MODELO DE ANALISE DE SENTIMENTOS ---")
for epoca in range(1, 151):
    previsoes = modelo_nlp(X_treino)
    perda = criterio(previsoes, y_treino)

    otimizador.zero_grad()
    perda.backward()
    otimizador.step()

    if epoca % 30 == 0:
        print(f"Época {epoca}/150 - Perda: {perda.item():.4f}")

# 5. Função para prever frases inéditas
def analisar_frase(frase):
    modelo_nlp.eval()
    with torch.no_grad():
        tokens = torch.tensor([frase_para_tokens(frase)], dtype=torch.long)
        prob = modelo_nlp(tokens).item()
        sentimento = "POSITIVO ✅" if prob >= 0.5 else "NEGATIVO ❌"
        print(f"Frase: '{frase}' | Probabilidade: {prob:.2%} -> {sentimento}")

print("\n--- TESTANDO FRASES NOVAS ---")
analisar_frase("produto muito bom e entrega rápida")
analisar_frase("odiei esse serviço horrível")
analisar_frase("comida excelente super recomendo")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

# 1. Preparação dos dados de imagem (Download e Transformação)
transformacao = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,)) # Normaliza o brilho dos pixels
])

# Baixando dataset MNIST (Dígitos de 0 a 9)
dados_treino = datasets.MNIST(root='./data', train=True, download=True, transform=transformacao)
dados_teste = datasets.MNIST(root='./data', train=False, download=True, transform=transformacao)

# Criando os DataLoaders (carregam as imagens em lotes/batches)
carregador_treino = torch.utils.data.DataLoader(dados_treino, batch_size=64, shuffle=True)
carregador_teste = torch.utils.data.DataLoader(dados_teste, batch_size=1000, shuffle=False)

# 2. Arquitetura da Rede Neural para Imagens
class ClassificadorImagens(nn.Module):
    def __init__(self):
        super(ClassificadorImagens, self).__init__()
        # A imagem do MNIST é 28x28 pixels = 784 entradas
        self.camada_entrada = nn.Linear(28 * 28, 128)
        self.relu = nn.ReLU()
        self.camada_saida = nn.Linear(128, 10) # 10 saídas (dígitos de 0 a 9)

    def forward(self, x):
        # Achatando a imagem 2D (28x28) para um vetor 1D (784)
        x = x.view(-1, 28 * 28)
        x = self.relu(self.camada_entrada(x))
        x = self.camada_saida(x)
        return x

modelo_visao = ClassificadorImagens()
criterio = nn.CrossEntropyLoss()
otimizador = optim.Adam(modelo_visao.parameters(), lr=0.001)

# 3. Treinamento da Rede em Imagens Reais
print("--- TREINANDO REDE DE VISÃO COMPUTACIONAL ---")
epochs = 3
for epoca in range(1, epochs + 1):
    modelo_visao.train()
    perda_total = 0
    for imagens, rotulos in carregador_treino:
        otimizador.zero_grad()
        saidas = modelo_visao(imagens)
        perda = criterio(saidas, rotulos)
        perda.backward()
        otimizador.step()
        perda_total += perda.item()

    print(f"Época {epoca}/{epochs} - Perda Média: {perda_total / len(carregador_treino):.4f}")

# 4. Testando a Precisão da Rede
modelo_visao.eval()
corretos = 0
total = 0

with torch.no_grad():
    for imagens, rotulos in carregador_teste:
        saidas = modelo_visao(imagens)
        _, previsoes = torch.max(saidas.data, 1)
        total += rotulos.size(0)
        corretos += (previsoes == rotulos).sum().item()

precisao = (corretos / total) * 100
print(f"\n--- RESULTADO FINAL ---")
print(f"Acurácia da Rede nas 10.000 imagens de teste: {precisao:.2f}% ✅")
```
